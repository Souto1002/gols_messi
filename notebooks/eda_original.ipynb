## Passo 1: Importação das Bibliotecas e Leitura dos Dados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('messi_all_goals.csv')

df.head()
```

## Passo 2: Inspeção e Diagnóstico Inicial dos Dados

```python
df.info()

df.isnull().sum()
```

## Passo 3: Pré-processamento e Limpeza de Dados

### A. Padronização de Nomes de Variáveis / Colunas

```python
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')
df.head(2)
```

**Justificativa:** A padronização dos nomes das variáveis para minúsculas e sem espaços/caracteres especiais facilita o manuseio das colunas no código, evitando erros de digitação durante a análise.

### B. Tratamento de Valores Ausentes e Exclusão de Linhas

```python
if 'assist' in df.columns:
    df['assist'] = df['assist'].fillna('Sem assistência')

df.isnull().sum()
```

**Justificativa para não exclusão de linhas:** Nenhuma linha do dataset foi excluída. A ausência de valor na coluna `assist` indica que o gol foi fruto de uma jogada individual ou sem assistência direta registrada, não sendo um dado faltante incorreto. A remoção de linhas resultaria em perda irreparável do histórico de gols do atleta.

### C. Remoção de Colunas Irrelevantes

**Justificativa:** Este passo **não precisou ser executado**. Nenhuma coluna foi removida do dataset original, pois todas as variáveis presentes (tais como adversário, competição, minuto, tipo de gol, etc.) possuem valor contextual direto para a análise descritiva do desempenho do jogador.

### D. Adequação dos Tipos de Dados

```python
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])

df.dtypes
```

**Justificativa:** A conversão de datas para o tipo `datetime` permite realizar agrupamentos temporais por ano, mês ou temporada nas análises descritivas.
